## **E-Commerce Sales Performance**


# Goal:
The goal of this project was to analyze e-commerce sales product data to see monthly trends on sales, evaluate the impact of customer product reviews on sale volume, and identify the best product categories in performance.


# Key Insights & Findings:
Overall business revenue peaks in October/sales_month_10, with 514,798 units, and it is at it's lowest in May/sales_month_6 with 487,194 units.

Toys is the most sold category, with 79,836 sales in Month 1. Home & Kitchen being the least sold category, with 57,844 sales in Month 1.

The scatter plot reveals that product reviews and sales have no correlation with each other. Products with 1.0 rating averaged 500 sales, while 5.0 ratings averaged sales with 442.


# Solutions:
Allocate more marketing budget and increase inventory stockpiles ahead of October's peak, while running promotional discount campaignes in May to clear out items that move slow.

Prioritize marketing real estate for high demand items like Toys, Books, and Sports. Consider revamping Home & Kitchen to address the low demand.

Since reviews don't affect sales, focus quality assurence teams on customer retention and long-term brand loyalty rather than assuming low ratings immediately kill baseline transactions.



```python
import pandas as pd
```

```python
df = pd.read_csv("archive.zip")

df.head()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df.columns
```

```python
month_columns = [f'sales_month_{i}' for i in range(1, 13)]

monthly_totals = df[month_columns].sum()

print("Total Sales by Month:")
print(monthly_totals)
```

```python
import matplotlib.pyplot as plot
month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

plot.figure(figsize=(10, 5))

plot.plot(month_names, monthly_totals.values, marker='o', color='darkorange', linewidth=2.5)

plot.title('Total E-Commerce Sales Revenue by Month', fontsize=14, fontweight='bold', pad=15)
plot.xlabel('Month of the Year', fontsize=12)
plot.ylabel('Total Sales Units / Value', fontsize=12)

plot.grid(True, linestyle='--', alpha=0.5)

plot.tight_layout()
plot.show()
```

```python
# Group by product category and sum up sales from month 1 to get a quick breakdown
category_sales = df.groupby('category')['sales_month_1'].sum().sort_values(ascending=False)

print("Top Product Categories (Month 1):")
print(category_sales)
```

```python
# Group by the star rating (review_score) and find the average sales
review_trends = df.groupby('review_score')['sales_month_1'].mean()

print("Average Sales by Review Score:")
print(review_trends)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize = (10, 7))

plt.scatter(df['review_score'], df['sales_month_1'], alpha=0.5, color='darkorange')

plt.title('Product Review Scores vs. Month 1 Sales', fontsize=14, fontweight='bold')
plt.xlabel('Review Score (1 to 5 Stars)', fontsize=12)
plt.ylabel('Sales Volume', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
category_sales = df.groupby('category')['sales_month_1'].sum()

category_sales_sorted = category_sales.sort_values(ascending=False)

print("Top-Selling Product Categories (Month 1):")
print(category_sales_sorted)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize = (10,7))

plt.barh(category_sales_sorted.index, category_sales_sorted.values, color = "darkorange")

plt.gca().invert_yaxis()

plt.title('Top-Selling Product Categories (Month 1)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Total Sales Units', fontsize=12)
plt.ylabel('Product Category', fontsize=12)

plt.grid(axis = 'x', linestyle = '--', alpha = 0.5, color = "black")

plt.tight_layout()
plt.show()
```
